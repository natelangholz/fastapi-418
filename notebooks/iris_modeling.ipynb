# Iris Modeling

A Jupyter notebook for exploratory analysis, linear modeling, and model export.


## Intro

This notebook shows exploratory data analysis and model creation using the iris dataset.


```python
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.linear_model import LinearRegression

iris = load_iris(as_frame=True)
data_iris = iris.frame.rename(columns={
    'sepal length (cm)': 'Sepal.Length',
    'sepal width (cm)': 'Sepal.Width',
    'petal length (cm)': 'Petal.Length',
    'petal width (cm)': 'Petal.Width'
})
data_iris.head()
```

## Exploratory Data Analysis

### Plot


```python
plt.figure(figsize=(6, 4))
plt.scatter(data_iris['Petal.Width'], data_iris['Petal.Length'])
plt.xlabel('Petal.Width')
plt.ylabel('Petal.Length')
plt.title('Petal Width vs Petal Length')
plt.show()
```

### Correlation


```python
data_iris['Petal.Width'].corr(data_iris['Petal.Length'])
```

## Modeling

### Create Models


```python
model_1 = LinearRegression()
model_1.fit(data_iris[['Petal.Width']], data_iris['Petal.Length'])
print('Intercept:', model_1.intercept_)
print('Coefficient:', model_1.coef_[0])
print('R^2:', model_1.score(data_iris[['Petal.Width']], data_iris['Petal.Length']))
```

```python
model_2 = LinearRegression()
model_2.fit(data_iris[['Petal.Width', 'Sepal.Length']], data_iris['Petal.Length'])
print('Intercept:', model_2.intercept_)
print('Coefficients:', dict(zip(['Petal.Width', 'Sepal.Length'], model_2.coef_)))
print('R^2:', model_2.score(data_iris[['Petal.Width', 'Sepal.Length']], data_iris['Petal.Length']))
```

### Save Model

Run this cell to create the model file used by the API.


```python
repo_root = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
output_path = repo_root / 'models' / 'model_2.pkl'
output_path.parent.mkdir(parents=True, exist_ok=True)
joblib.dump(model_2, output_path)
print(f'Saved model to {output_path}')
```
